# Feature Engineering

gross margin

```python
df['Gross Margin %'] = (
    df['Gross Profit'] / df['Sales']
) * 100
```

Profit Per Unit

```python
df['Profit Per Unit'] = (
    df['Gross Profit'] / df['Units']
)
```

Revenue Contribution %

```python
total_sales = df['Sales'].sum()

df['Revenue Contribution %'] = (
    df['Sales'] / total_sales
) * 100
```

Profit Contribution %

```python
total_profit = df['Gross Profit'].sum()

df['Profit Contribution %'] = (
    df['Gross Profit'] / total_profit
) * 100
```

# Product Profitability Analysis

Top Profitable Products

```python
top_profit_products = (
    df.groupby('Product Name')['Gross Profit']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_profit_products
```

Visualization

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_profit_products.values,
    y=top_profit_products.index
)

plt.title("Top 10 Profitable Products")

plt.xlabel("Gross Profit")

plt.ylabel("Product Name")

plt.show()
```

Highest Margin Products

```python
top_margin_products = (
    df.groupby('Product Name')['Gross Margin %']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

top_margin_products
```

Margin Visualization

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_margin_products.values,
    y=top_margin_products.index
)

plt.title("Top Margin Products")

plt.xlabel("Average Margin %")

plt.ylabel("Product Name")

plt.show()
```


High Sales but Low Margin Products

```python
high_sales_low_margin = df[
    (df['Sales'] > df['Sales'].quantile(0.75)) &
    (df['Gross Margin %'] < df['Gross Margin %'].quantile(0.25))
]

high_sales_low_margin[
    ['Product Name', 'Sales', 'Gross Margin %']
].head()
```

Low Sales Low Profit Products

```python
low_sales_low_profit = df[
    (df['Sales'] < df['Sales'].quantile(0.25)) &
    (df['Gross Profit'] < df['Gross Profit'].quantile(0.25))
]

low_sales_low_profit[
    ['Product Name', 'Sales', 'Gross Profit']
].head()
```

Product Classification

```python
conditions = [
    
    (
        (df['Sales'] >= df['Sales'].median()) &
        (df['Gross Margin %'] >= df['Gross Margin %'].median())
    ),
    
    (
        (df['Sales'] >= df['Sales'].median()) &
        (df['Gross Margin %'] < df['Gross Margin %'].median())
    ),
    
    (
        (df['Sales'] < df['Sales'].median()) &
        (df['Gross Margin %'] >= df['Gross Margin %'].median())
    )
]

categories = [
    "High Sales High Profit",
    "High Sales Low Margin",
    "Low Sales High Margin"
]

df['Product Category'] = np.select(
    conditions,
    categories,
    default="Low Sales Low Profit"
)
```

Product Category Distribution

```python
plt.figure(figsize=(10,6))

sns.countplot(
    y='Product Category',
    data=df
)

plt.title("Product Category Distribution")

plt.show()
```

Division Level Performance Analysis

```python
division_analysis = (
    df.groupby('Division')
    .agg({
        'Sales':'sum',
        'Gross Profit':'sum',
        'Gross Margin %':'mean'
    })
    .reset_index()
)

division_analysis
```

Revenue vs Profit Comparison

```python
division_analysis.plot(
    x='Division',
    y=['Sales', 'Gross Profit'],
    kind='bar',
    figsize=(10,6)
)

plt.title("Division Revenue vs Profit")

plt.ylabel("Amount")

plt.show()
```

Margin Distribution by Division

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x='Division',
    y='Gross Margin %',
    data=df
)

plt.title("Margin Distribution by Division")

plt.show()
```

Pareto Analysis (80/20 Rule)

```python
pareto_df = (
    df.groupby('Product Name')['Gross Profit']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

pareto_df['Cumulative Profit'] = (
    pareto_df['Gross Profit'].cumsum()
)

pareto_df['Cumulative Percentage'] = (
    pareto_df['Cumulative Profit']
    /
    pareto_df['Gross Profit'].sum()
) * 100
```

Pareto Chart

```python
plt.figure(figsize=(14,6))

plt.plot(
    pareto_df.index,
    pareto_df['Cumulative Percentage'],
    marker='o'
)

plt.axhline(
    80,
    color='red',
    linestyle='--'
)

plt.title("Pareto Analysis - Profit Contribution")

plt.xlabel("Products")

plt.ylabel("Cumulative Profit %")

plt.show()
```

Margin Risk Detection

```python
margin_threshold = 10

df['Margin Risk'] = np.where(
    df['Gross Margin %'] < margin_threshold,
    'Risk',
    'Safe'
)
```

Risk Products

```python
risk_products = df[
    df['Margin Risk'] == 'Risk'
]

risk_products[
    ['Product Name', 'Gross Margin %']
].head()
```

# Margin Volatility Analysis

create monthly column

```python
df['Month'] = (
    df['Order Date']
    .dt
    .to_period('M')
)
```

Monthly Margin Trend

```python
monthly_margin = (
    df.groupby('Month')['Gross Margin %']
    .mean()
)
```

Margin Volatility Plot

```python
monthly_margin.plot(
    figsize=(12,6)
)

plt.title("Monthly Margin Volatility")

plt.ylabel("Average Margin %")

plt.show()
```

Correlation Analysis
